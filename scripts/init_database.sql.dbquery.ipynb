```sql
/*
=============================================================
Create Database and Schemas
=============================================================
Script Purpose:
    This script creates a new catalog named 'DataWarehouse'.Additionally, the script sets up three schemas 
    within the database: 'bronze', 'silver', and 'gold'.
*/

-- Create the 'DataWarehouse' catalog
CREATE CATALOG IF NOT EXISTS DataWarehouse;

USE CATALOG DataWarehouse;

-- Create Schemas
CREATE SCHEMA bronze;

CREATE SCHEMA silver;

CREATE SCHEMA gold;
```
